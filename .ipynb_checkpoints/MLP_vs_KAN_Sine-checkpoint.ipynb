# Learning a sine function with an MLP and a KAN

Both models learn $y=\sin(x)$ over $[-2\pi, 2\pi]$ from the same 64 labeled points.
They each have 34 trainable parameters and use full-batch Adam with the same learning rate and update budget.
Follow their predictions through training and compare the errors at unseen points between training samples.

Run the notebook from top to bottom with a Python kernel that has PyTorch and Matplotlib installed. A CPU is enough.

```python
import math

import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.nn import functional as F

torch.set_num_threads(1)

SEED = 7
N_TRAIN = 64
TEST_POINTS_PER_GAP = 8
CHECKPOINTS = (0, 50, 300, 2000)
LEARNING_RATE = 0.01

x_train = torch.linspace(-2 * math.pi, 2 * math.pi, N_TRAIN).reshape(-1, 1)
fractions = (torch.arange(TEST_POINTS_PER_GAP) + 0.5) / TEST_POINTS_PER_GAP
x_test = (x_train[:-1] + x_train.diff(dim=0) * fractions).reshape(-1, 1)
x_plot = torch.linspace(-2 * math.pi, 2 * math.pi, 600).reshape(-1, 1)

y_train, y_test, y_plot = (torch.sin(x) for x in (x_train, x_test, x_plot))

# Both models receive the same normalized input u in [-1, 1].
u_train, u_test, u_plot = (x / (2 * math.pi) for x in (x_train, x_test, x_plot))
```

## How the models represent the curve

**MLP.** Each of 11 hidden units applies the same fixed $\tanh$ function. Training learns the input weights $w_j$, hidden biases $b_j$, output weights $a_j$, and output bias $c$.
$$\hat y(u) = c + \sum_{j=1}^{11} a_j\tanh(w_j u + b_j)$$

**KAN.** A single learnable edge connects the scalar input to the scalar output. The edge combines a SiLU base term with 33 cubic B-spline basis functions $B_k$. Training learns $\alpha$ and the coefficients $\beta_k$. The basis functions are centered on a fixed grid, so each coefficient can change the curve in a local region.
$$\hat y(u) = \phi(u) = \alpha\,\mathrm{SiLU}(u) + \sum_{k=1}^{33}\beta_k B_k(u)$$

This one-edge KAN is sufficient to demonstrate spline learning for a scalar input and output. [Liu et al., *KAN: Kolmogorov–Arnold Networks*](https://arxiv.org/abs/2404.19756)

```python
def cubic_spline_basis(z):
    # Centered cubic B-spline with support on [-2, 2].
    distance = z.abs()
    inner = 2 / 3 - distance.square() + 0.5 * distance.pow(3)
    outer = (2 - distance).clamp_min(0).pow(3) / 6
    return torch.where(distance < 1, inner, outer)


class SplineKAN(nn.Module):
    def __init__(self):
        super().__init__()
        # A uniform grid with one extra center on each side of [-1, 1].
        self.spacing = 1 / 15
        self.register_buffer("centers", torch.arange(-16, 17) * self.spacing)
        self.coefficients = nn.Parameter(torch.randn(self.centers.numel()) * 0.02)
        self.base_weight = nn.Parameter(torch.zeros(()))

    def forward(self, u):
        basis = cubic_spline_basis((u - self.centers) / self.spacing)
        spline = basis @ self.coefficients[:, None]
        return self.base_weight * F.silu(u) + spline


torch.manual_seed(SEED)
models = {
    "MLP": nn.Sequential(nn.Linear(1, 11), nn.Tanh(), nn.Linear(11, 1)),
    "KAN": SplineKAN(),
}
for name, model in models.items():
    parameters = sum(parameter.numel() for parameter in model.parameters())
    print(f"{name}  {parameters} trainable parameters")
```

## Train on the same samples

Update 0 captures the initial prediction. After each update, record mean squared error (MSE) on the training samples and the interleaved test grid. Only the training error is used for backpropagation.

```python
def train(model):
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)
    history = {"train": [], "test": []}
    predictions = {}

    @torch.no_grad()
    def record(step):
        history["train"].append(F.mse_loss(model(u_train), y_train).item())
        history["test"].append(F.mse_loss(model(u_test), y_test).item())
        if step in CHECKPOINTS:
            predictions[step] = model(u_plot).squeeze()

    record(0)
    for step in range(1, CHECKPOINTS[-1] + 1):
        optimizer.zero_grad()
        loss = F.mse_loss(model(u_train), y_train)
        loss.backward()
        optimizer.step()
        record(step)

    return history, predictions


results = {name: train(model) for name, model in models.items()}
```

## Watch the fits change

The black line is the exact sine function. Each panel shows both predictions after the indicated number of optimizer updates.

```python
colors = {"MLP": "#0072B2", "KAN": "#D55E00"}
styles = {"MLP": "--", "KAN": "-."}

fig, axes = plt.subplots(2, 2, figsize=(10, 6), sharex=True, sharey=True,
                         layout="constrained")
for ax, step in zip(axes.flat, CHECKPOINTS):
    ax.plot(x_plot, y_plot, color="black", alpha=0.7, label="True sine")
    ax.scatter(x_train, y_train, s=10, color="0.6", label="Training samples")
    for name, (_, predictions) in results.items():
        ax.plot(x_plot, predictions[step], styles[name], color=colors[name],
                linewidth=2, label=name)
    ax.set_title(f"Update {step:,}")
    ax.grid(alpha=0.2)

axes[0, 0].legend(ncol=2, fontsize=9)
fig.supxlabel("x (radians)")
fig.supylabel("y")
plt.show()
```

## Compare error and the learned spline

The left panel tracks error on unseen points. The right panel shows how the KAN's edge builds the final prediction from its learned spline and scaled SiLU. The test grid measures interpolation across the sampled interval.

```python
fig, (loss_ax, edge_ax) = plt.subplots(1, 2, figsize=(11, 4), layout="constrained")

for name, (history, _) in results.items():
    loss_ax.semilogy(history["test"], color=colors[name], label=name)
loss_ax.set(xlabel="Optimizer updates", ylabel="Test MSE", title="Learning curves")
loss_ax.legend()

kan = models["KAN"]
with torch.no_grad():
    base = kan.base_weight * F.silu(u_plot)
    edge = kan(u_plot)
edge_ax.plot(x_plot, edge - base, color="#009E73", label="Learned spline")
edge_ax.plot(x_plot, base, "--", color="0.4", label="Scaled SiLU")
edge_ax.plot(x_plot, edge, color=colors["KAN"], label="Sum (KAN prediction)")
edge_ax.set(xlabel="x (radians)", ylabel="Edge output", title="Final KAN edge")
edge_ax.legend()

for ax in (loss_ax, edge_ax):
    ax.grid(alpha=0.2)
plt.show()

print(f"{'Model':<8} {'Train MSE':>12} {'Test MSE':>12}")
for name, (history, _) in results.items():
    print(f"{name:<8} {history['train'][-1]:>12.3e} {history['test'][-1]:>12.3e}")
```

Each run compares one initialization of two small models. The fixed spline grid gives the KAN a local basis for this smooth one-dimensional function. The MLP learns the positions, scales, and weights of its tanh units. The relative errors depend on the architecture, grid, and training settings.
